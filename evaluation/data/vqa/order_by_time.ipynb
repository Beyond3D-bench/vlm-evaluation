```python
import json
import pandas as pd

input_path = "/work/courses/3dv/team1/data/vqa/visible/selected_201_250.jsonl"
output_path = "/work/courses/3dv/team1/data/vqa/visible/selected_201_250_sorted.jsonl"

df = pd.read_json(input_path, lines=True)

df["query_time_sec"] = pd.to_numeric(df["query_time_sec"], errors="coerce")

df_sorted = df.sort_values(
    by="query_time_sec",
    ascending=True,
    na_position="last"
)

with open(output_path, "w", encoding="utf-8") as f:
    for record in df_sorted.to_dict(orient="records"):
        f.write(json.dumps(record, ensure_ascii=False) + "\n")
```
